```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


# data 

data = {
    "screen_time":[8,3,7,2,9,4,6,2,8,5],
    "night_used":[1,0,1,0,0,1,0,1,0,1],
    "sleeping_hr":[5,8,6,9,5,8,6,8,5,5],
    "addict":[1,0,1,0,1,0,1,0,0,1]
}
df = pd.DataFrame(data)

#Work data

x = df[["screen_time","night_used","sleeping_hr"]]
y = df["addict"]
# model_creat
model = RandomForestClassifier()
#train_model
model.fit(x,y)
pre = model.predict(x)
# score 
score = accuracy_score(pre,y)
# print(score)

new_data = [[screen_time,nigh_use,sleep]]
result = model.predict(new_data)
print(result)

screen_time = int(input("enter_value_screen_time"))
nigh_use = int(input("enter_vale_night_use"))
sleep = int(input("enter_value_sleep"))
```
