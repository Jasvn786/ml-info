```python
import streamlit as st
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score


# data 

data = {
    "screen_time":[8,3,7,2,9,4,6,2,8,5],
    "night_used":[1,0,1,0,0,1,0,1,0,1],
    "sleeping_hr":[5,8,6,9,5,8,6,8,5,5],
    "addict":[1,0,1,0,1,0,1,0,0,1]
}
df = pd.DataFrame(data)

#Work data

x = df[["screen_time","night_used","sleeping_hr"]]
y = df["addict"]
# model_creat
model = RandomForestClassifier(n_estimators=100, random_state=42)
#train_model
model.fit(x,y)
pre = model.predict(x)
# score 
score = accuracy_score(pre,y)
# print(score)

screen_time = st.number_input("enter_value_screen_time", min_value=0)
nigh_use = st.number_input("enter_vale_night_use", min_value=0, max_value=1)
sleep = st.number_input("enter_value_sleep", min_value=0)

new_data = [[screen_time,nigh_use,sleep]]
result = model.predict(new_data)
print(result)

if result[0] == 1:
    print("end=> you are addicted to mobile")
```
