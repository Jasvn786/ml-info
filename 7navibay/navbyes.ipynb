```python
# import libary
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import KFold,cross_val_score
from sklearn.pipeline import Pipeline

# Data
email = ["win money now", 
"free lottery",
"meeting schedule",
"claim your money",
"invite you sir",
"love you you dear"]

labels = ["spam","spam","normal","spam","normal","normal"]
y = labels

# text to number becuse system olny need 0 1 
# pipline

"""model = pipeline([(

)])"""

vector = TfidfVectorizer()
x = vector.fit_transform(email)

# creted model
model = MultinomialNB()

kf = KFold(n_splits= 5 , shuffle= True)
scor = cross_val_score(model , x , y,cv = kf)

print(scor.mean())  


# train model 

model.fit(x,labels)
# chek model 


new_data = vector.transform(["love you dear and give some money and win more money beliwe me "])
prediction = model.predict(new_data)
print(prediction)
```
